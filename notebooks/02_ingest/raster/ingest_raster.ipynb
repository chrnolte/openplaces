**`ingest_raster`**

Script examples to import a raster dataset

# Configure

```python
import argparse

from openplaces.io.ingester import Ingester
from openplaces.recipe import get_recipe_by_id
from openplaces.utils import pretty_print
```

```python
# Define arguments
parser = argparse.ArgumentParser(description='Ingest rasters using a recipe')
parser.add_argument(
    '--recipe_id',
    help='Identifier of the recipe (e.g., "US_terrain-usgs-3dep" or '
    '"US_landcover-annual-lcms-v202410")',
)
parser.add_argument(
    '--admin_ids',
    help='Administrative unit IDs to ingest (e.g., "US-RI")',
    nargs='*',
)
parser.add_argument(
    '--partition_ids',
    help='Partitions to ingest (e.g., "2024")',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess input data from downloaded file',
    action='store_true',
)
parser.add_argument(
    '--redownload',
    help='Redownload input data from original source',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    help='If True, print outputs while processing data',
    action='store_true',
)
```

# Set arguments

```python
ARGS_TEST = (
    # First build-up year: US-HISDAC
    '--recipe_id US_built-year-hisdac-v1 '
    # Elevation: USGS 3DEP (~10m, EPSG:4326, per county)
    # '--recipe_id US_land-elevation-usgs-3dep '
    # Land cover change: USFS LCMS, annual
    # '--recipe_id US_landcover-change-lcms-202410 '
    # '--partition_ids 2000 2024 '
    # Raster recipe #3: USFS annual land cover (LCMS)
    # '--recipe_id US_landcover-annual-lcms-202410 '
    # '--partition_ids 2000 2024 '
    # Location #1: Newport county, RI (fastest US county)
    '--admin_ids US-RI-NE '
    # Location #2: Brunswick and Buncombe, NC (flood/hurricane risk cases)
    # '--admin_ids US-NC-BR US-NC-BU '
    # Location #3: Harris and Jefferson, TX (hurricane risk cases)
    # '--admin_ids US-TX-HAR US-TX-JEF '
    # Location #4: Oneida and Vilas counties, WI (lake hedonics case study)
    # '--admin_ids US-WI-ON US-WI-VI'
    # Location #5: Hillsborough, Pinellas, and Manatee counties, FL (Tampa bay)
    # '--admin_ids US-FL-HI US-FL-MN US-FL-PI'
    # Location #6: Eastern North Carolina (CHEER building inventory)
    # '--admin_ids US-NC-BE US-NC-BT US-NC-BL US-NC-BR US-NC-CM US-NC-AR US-NC-CW US-NC-CO US-NC-CR US-NC-CU US-NC-CI US-NC-DR US-NC-DP US-NC-ED US-NC-FR US-NC-GT US-NC-GE US-NC-HL US-NC-HA US-NC-HR US-NC-HO US-NC-HD US-NC-JO US-NC-JN US-NC-LN US-NC-MR US-NC-NA US-NC-NH US-NC-NO US-NC-ON US-NC-PM US-NC-PA US-NC-PE US-NC-PQ US-NC-PI US-NC-RB US-NC-SA US-NC-SC US-NC-TY US-NC-WA US-NC-WR US-NC-WS US-NC-WY US-NC-WI'
    '--reprocess '
    # '--redownload '
    '--verbose '
)

# Convert argument string to list of strings
args_list = [x for x in ARGS_TEST.split(' ') if x != '']

# Parse list of arguments
args = parser.parse_args(args_list)

# Display arguments to check if parsing worked as expected
args
```

# Show recipe

```python
pretty_print(get_recipe_by_id(args.recipe_id))
```

# Ingest

```python
ingester = Ingester(
    args.recipe_id,
    args.admin_ids,
    args.partition_ids,
    verbose=args.verbose,
)
```

```python
ingester.ingest(reprocess=args.reprocess, redownload=args.redownload)
```
