**`ingest_dwellings`**

Script examples to import dwelling / address datasets

# Configure

```python
import argparse

from openplaces.io.ingester import Ingester
```

```python
# Define arguments
parser = argparse.ArgumentParser(description='Ingest dwellings using a recipe')
parser.add_argument(
    '--recipe_id',
    help='Identifier of the recipe (e.g., "dwelling-overture-2025")',
)
parser.add_argument(
    '--admin_ids',
    help='Administrative unit IDs to ingest (e.g., "US-RI-NE")',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess input data from downloaded file',
    action='store_true',
)
parser.add_argument(
    '--redownload',
    help='Redownload input data from original source',
    action='store_true',
)
parser.add_argument(
    '--keep_unzipped',
    help='If True, keeps unzipped datasets in heap folder after processing',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    help='If True, print outputs while processing data',
    action='store_true',
)
```

# Test arguments

```python
ARGS_TEST = (
    # Global: Overture addresses
    '--recipe_id dwelling-overture-2025 '
    # Newport county, RI (fastest US county)
    '--admin_ids US-RI-NE '
    # Brunswick and Buncombe, NC (flood/hurricane risk cases)
    # '--admin_ids US-NC-BR US-NC-BU '
    # Harris and Jefferson, TX (hurricane risk cases)
    # '--admin_ids US-TX-HAR US-TX-JEF '
    # Processing flags
    '--reprocess '
    # '--redownload '
    '--verbose '
    # '--keep_unzipped '
)

args_list = [x for x in ARGS_TEST.split(' ') if x != '']
args = parser.parse_args(args_list)
args
```

```python
from openplaces.recipe import get_recipe_by_id
from openplaces.utils import pretty_print

pretty_print(get_recipe_by_id(args.recipe_id))
```

# Ingest dwelling data

```python
ingester = Ingester(args.recipe_id, args.admin_ids, verbose=args.verbose)
```

```python
ingester.ingest(
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
)
```
