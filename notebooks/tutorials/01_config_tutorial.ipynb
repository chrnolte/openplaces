**`01_config_tutorial`**

A tutorial for configuring `openplaces`.

# Initialize configuration
The first use of `import openplaces` triggers an interactive configuration setup.

The setup script creates a user configuration file in the user application folder.

Setup will initiate whenever that file does not exist.

```python
import openplaces
```

# Access attributes

```python
cfg = openplaces.cfg
```

## Access directories

```python
cfg.code_root
```

```python
cfg.data_root
```

```python
cfg.core_dir
```

```python
cfg.models_dir
```

## Other attributes
Usually defined at project level (`openplaces.yaml`), but users can override in personal config file

```python
cfg.geo_ha_min
```

## Entire configuration

```python
from openplaces.config import show_config
show_config()
```

# Edit configuration file
Opens local text editor

```python
from openplaces.config import edit_config
edit_config()
```

```python
from openplaces.config import reload_config
reload_config()
```

# Reset configuration
Restart kernel afterwards to reset `cfg` and re-initialize the configuration

```python
from openplaces.config import reset_config
reset_config()
```
