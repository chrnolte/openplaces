**`01_show_administrative_units`**

Tutorial for the API of admin units.

```python
# Map making
import matplotlib.pyplot as plt

# Data access for administrative units
from openplaces.api import get_admin
```

Maps draw the boundaries on a plain background.

- The selected unit is black.
- Its parent units are gray.

# Countries, territories, etc.

```python
# Get admin units. Default level is 1 (countries).
admin1 = get_admin()
admin1.sample(5).sort_index()
```

```python
# Map of random country / territory
random_admin1_id = admin1.sample().index[0]
fig, ax = plt.subplots(figsize=(8, 8))
# Pull geometry
random_admin1 = get_admin(random_admin1_id, geom=True)
ax = random_admin1.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(random_admin1['name'].iloc[0] + f' ({random_admin1_id})')
```

# States, departments, etc.

```python
admin2 = get_admin(random_admin1_id, level=2, geom=True)
admin2.sample(min(len(admin2), 5)).sort_index()
```

```python
# Map of random country / territory
random_admin2_id = admin2.sample().index[0]
random_admin2 = get_admin(random_admin2_id, geom=True)

admin2_label = (
    random_admin2['name'].iloc[0]
    + ', '
    + random_admin1['name'].iloc[0]
    + f' ({random_admin2_id})'
)

fig, ax = plt.subplots(figsize=(7, 7))

random_admin1.boundary.plot(ax=ax, color='gray', linewidth=0.5)
random_admin2.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(admin2_label)

fig, ax = plt.subplots(figsize=(10, 10))
random_admin2.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(admin2_label)
```

# Counties, districts, etc.

```python
admin3 = get_admin(random_admin2_id, level=3, geom=True)
admin3.sample(min(len(admin3), 5)).sort_index()
```

```python
# Map of random county, district, etc.
random_admin3_id = admin3.sample().index[0]
random_admin3 = get_admin(random_admin3_id, geom=True)

admin3_label = (
    random_admin3['name'].iloc[0]
    + ', '
    + random_admin2['name'].iloc[0]
    + ', '
    + random_admin1['name'].iloc[0]
    + f' ({random_admin3_id})'
)

fig, ax = plt.subplots(figsize=(7, 7))
random_admin1.boundary.plot(ax=ax, color='gray', linewidth=0.5)
random_admin2.boundary.plot(ax=ax, color='gray', linewidth=1)
random_admin3.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(admin3_label)

fig, ax = plt.subplots(figsize=(7, 7))
random_admin2.boundary.plot(ax=ax, color='gray', linewidth=1)
random_admin3.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(admin3_label)

fig, ax = plt.subplots(figsize=(14, 14))
random_admin3.boundary.plot(ax=ax, color='black', linewidth=1)
ax.set_title(admin3_label)
```
