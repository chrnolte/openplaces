```python
# Get path to file in the US-wide data directory (adds prefix to file)
path('US', filename='data.csv')
```

```python
# File in Massachusetts (with prefix)
path('US-MA', filename='data.csv')
```

```python
# Directory of parcel data in Massachusetts
path('US-MA', 'parcel-massgis-2018')
```

```python
# Path to intersected data in `openplaces` data folder
# - For United States > Massachusetts > Michigan
# - Entities: parcels, as per MassGIS (state service), in 2018
# - Features: water > lakes data > from the National Hydrography Dataset > High-Res
# Adds file prefixes and `.parquet` extension by default
path('US-MA-MI', 'parcel-massgis-2018', 'water-lake-nhd-2019hr')
```

```python
# Path to intersected data in `openplaces` core data folder
# - Where: United States > California
# - Entities: buildings, as per Microsoft footprints, 2019
# - Features: landcover > forest > from Hansen's dataset, 2024 update
# Adds file prefixes and `.parquet` extension by default
path(
    admin_id='US-CA',
    entity='building-microsoft-2019',
    dataset='landcover-forest-hansen-2024',
)
```

```python
# Path to downloaded data in `openplaces` external data folder
# (the default for downloads - without file prefixes)
external_path(
    admin_id='US-ND',
    dataset='water-lake-nhd-2019hr',
    filename='NHD_H_North_Dakota_State_GDB.zip',
)
```

# Filepath components

## Admin ID

```python
from openplaces.path import AdminId
```

```python
# Global
admin_id = AdminId()
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

```python
# Country
admin_id = AdminId('US')
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

```python
# State, department
admin_id = AdminId('US', 'MA')
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

```python
# State, department
admin_id = AdminId('US-MA')
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

```python
# County, municipality
admin_id = AdminId('US-MA-MI')
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

```python
# Town, etc.
admin_id = AdminId('US-MA-MI-SO')
print('ID:          ', str(admin_id))
print('Path:        ', admin_id.to_path())
print('File prefix: ', admin_id.to_prefix())
```

## Entity

```python
from openplaces.path import Entity, Source
```

```python
source = Source(
    source_id='massgis',
    url='https://www.mass.gov/info-details/massgis-data-property-tax-parcels',
)
str(source)
```

```python
# Create an entity without a version: version defaults to date
entity = Entity('parcel', source)
entity.to_path()
```

```python
# Create an entity with a named version
entity = Entity('parcel', source, '2018')
entity.to_path()
```
