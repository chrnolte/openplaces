**`01_first_steps`**

Start here and make it yours.

Here is how to set up ``openplaces`` for your system.

We'll start by setting the directory structure for data, models, and outputs.

This user / project configuration can later be accessed via ``openplaces.config.cfg``.

# Set up openplaces directories
The first use of `import openplaces` triggers an interactive configuration setup.

After completion, it will leave a user configuration file in your user application folder (location varies by operating system).

Setup will re-initiate whenever that file does not exist.

```python
import openplaces
```

# Access attributes

```python
from openplaces.config import cfg
```

```python
# Location of user configuration file
cfg.user_config_path
```

```python
from openplaces.config import show_config
show_config()
```

## Access directories

```python
cfg.code_root
```

```python
cfg.data_root
```

```python
cfg.core_dir
```

```python
cfg.external_dir
```

```python
cfg.models_dir
```

## Other attributes
Defined in library `openplaces.core.CONSTANTS`, for project (`openplaces.yaml`), or in user configuration file.

```python
# Smallest polygon area (in hectares) retained when importing polygon
# (areas smaller than one square meter are usually slivers)
cfg.geo_min_area_m2
```

Opens local text editor

# Reset configuration
Restart kernel afterwards to reset `cfg` and re-initialize the configuration

```python
from openplaces.config import reset_config
reset_config()
```
