**`01_first_steps`**

This notebook assists you in configuring ``openplaces`` for your system.

The first step is to define where your data will be stored.

This involves defining separate directories for data inputs, interim/cached data, core data, output data, models, reports, and processing logs.

Having separate directories simplifies sharing data and setting permissions, especially in multi-user setups (e.g. on a cluster).

# Set up directories
The first use of `import openplaces` triggers an interactive configuration setup.

After completion, the configuration will be saved as a file to your local user directory:
- On Windows: ``%APPDATA%\openplaces\config.yaml``
- On Mac: ``~/Library/Application Support/openplaces/config.yaml``
- On Linux: ``~/.config/openplaces/config.yaml``

Setup will re-initiate whenever that file does not exist.

```python
import openplaces
```
