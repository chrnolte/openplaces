**`enrich_footprints_with_image_detection`**

Recipe-driven footprint enrichment from satellite and street-level imagery.

# Configure

```python
import argparse

from openplaces.io.enricher import Enricher
from openplaces.recipe import get_output_path, get_recipe_by_id
from openplaces.utils import pretty_print
```

```python
parser = argparse.ArgumentParser(
    description='Enrich harmonized footprints with image detection'
)
parser.add_argument(
    '--recipe_ids',
    help='Image enrichment recipe IDs',
    nargs='+',
)
parser.add_argument(
    '--entity_recipe_id',
    help='Optional concrete footprint recipe override',
)
parser.add_argument(
    '--admin_ids',
    help='Admin unit IDs to enrich (e.g. "US-FL-AL")',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess admin IDs even if output already exists',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    action='store_true',
)
```

# Test arguments

```python
ARGS_TEST = (
    '--recipe_ids '
    'US_footprint_built-roof-shape-brails-2026 '
    'US_footprint_built-occupancy-brails-2026 '
    'US_footprint_built-n-stories-brails-2026 '
    '--entity_recipe_id US_footprint-spine-2026 '
    '--admin_ids US-FL-AL '
    '--reprocess '
    '--verbose'
)

args_list = [x for x in ARGS_TEST.split(' ') if x]
args = parser.parse_args(args_list)
args
```

```python
for recipe_id in args.recipe_ids:
    pretty_print(get_recipe_by_id(recipe_id))
```

# Run

```python
for recipe_id in args.recipe_ids:
    enricher = Enricher(
        recipe_id,
        admin_ids=args.admin_ids,
        entity_recipe_id=args.entity_recipe_id,
        verbose=args.verbose,
    )
    enricher.enrich(reprocess=args.reprocess)
```
