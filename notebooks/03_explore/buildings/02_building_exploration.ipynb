```python
from openplaces.api import read_entities
from openplaces.viz import show_building
```

# Inspect

```python
admin_id = 'US-NC-BR'  # Brunswick, NC (CHEER pilot)
```

```python
parcels = read_entities(admin_id, 'US-NC_parcel-nconemap-2025', geom=True)
buildings_fema = read_entities(admin_id, 'US_building-fema-2023', geom=True)
buildings_nsi = read_entities(admin_id, 'US_building-usace-2022', geom=True)
buildings_microsoft = read_entities(admin_id, 'US_building-microsoft-v2', geom=True)
```

```python
buildings_fema_sample = buildings_fema.sample(1)
```

```python
show_building(
    location=buildings_fema_sample,
    geodatasets={
        'parcels': parcels,
        'buildings_fema': buildings_fema,
        'buildings_nsi': buildings_nsi,
        'buildings_microsoft': buildings_microsoft,
    },
)
```
