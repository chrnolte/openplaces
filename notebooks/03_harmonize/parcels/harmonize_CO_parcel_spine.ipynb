**`harmonize_CO_parcel_spine`**

Recipe-driven harmonization of the Colombia parcel spine using `Harmonizer`.

# Configure

```python
import argparse

from openplaces.io.harmonizer import Harmonizer
```

```python
parser = argparse.ArgumentParser(description='Harmonization of the Colombia parcel spine')
parser.add_argument(
    '--recipe_id',
    help='Harmonization recipe',
)
parser.add_argument(
    '--admin_ids',
    help='Admin unit IDs to harmonize (e.g. "CO-AN-AD"). Omit to process all of Colombia.',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess admin IDs even if output already exists',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    action='store_true',
)
```

# Test arguments

```python
ARGS_TEST = (
    '--recipe_id CO_parcel-spine-2026 '
    # A municipality only IGAC covers, and one only the Antioquia catastro covers.
    '--admin_ids CO-TO-CY CO-AN-AD '
    '--reprocess '
    '--verbose'
)

args_list = [x for x in ARGS_TEST.split(' ') if x]
args = parser.parse_args(args_list)
args
```

```python
# Show recipe parameters
from openplaces.recipe import get_recipe_by_id
from openplaces.utils import pretty_print

pretty_print(get_recipe_by_id(args.recipe_id))
```

# Run

```python
harmonizer = Harmonizer(args.recipe_id, args.admin_ids, verbose=args.verbose)
```

```python
harmonizer.harmonize(reprocess=args.reprocess)
```

# Inspect results

```python
import openplaces as op

for admin_id in args.admin_ids:
    gdf = op.get_entities(args.recipe_id, admin_id, geom=True)
    print(admin_id, gdf.shape)
    print(gdf['geometry_source'].value_counts())
    print()
```
