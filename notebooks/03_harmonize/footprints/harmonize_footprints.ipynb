**`harmonize_footprints`**

Ingests all precursor datasets and runs recipe-driven footprint harmonization.

See `src/openplaces/recipes/US/_all/footprint/cheer/2026/US_footprint-cheer-2026.yaml`
for all configurable thresholds.

# Configure

```python
import argparse

from openplaces.io.harmonizer import harmonize
from openplaces.io.ingester import ingest
```

```python
parser = argparse.ArgumentParser(
    description='Ingest precursors and harmonize footprints'
)
parser.add_argument(
    '--recipe_id',
    help='Harmonization recipe (e.g. "US_footprint-cheer-2026")',
)
parser.add_argument(
    '--admin_ids',
    help='Admin unit IDs to process (e.g. "US-NC-BS")',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess admin IDs even if output already exists',
    action='store_true',
)
parser.add_argument(
    '--redownload',
    help='Redownload input data from original source',
    action='store_true',
)
parser.add_argument(
    '--keep_unzipped',
    help='Keep unzipped datasets in heap folder after processing',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    action='store_true',
)
```

# Test parameters

```python
ARGS_TEST = (
    '--recipe_id US_footprint-cheer-2026 '
    # Brunswick, NC (CHEER pilot: hurricane/flood risk)
    '--admin_ids US-NC-BS '
    # '--admin_ids US-FL-AL '
    # '--admin_ids US-TX-JE '
    '--reprocess '
    # '--redownload '
    # '--keep_unzipped '
    '--verbose'
)

args_list = [x for x in ARGS_TEST.split(' ') if x]
args = parser.parse_args(args_list)
args
```

```python
from openplaces.recipe import get_recipe_by_id
from openplaces.utils import pretty_print

pretty_print(get_recipe_by_id(args.recipe_id))
```

## Ingest precursor datasets

```python
# Global building footprints: OpenBuildingsMap (OBM)
ingest(
    'footprint-obm-2025',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# US building footprints: Microsoft
ingest(
    'US_footprint-microsoft-v2',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# State building footprints (auto-discovered by the pipeline).
# Change recipe_id to match the target state; omit if no state recipe exists.
# North Carolina: US-NC_footprint-ncdps-2023
ingest(
    'US-NC_footprint-ncdps-2023',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# US building footprints: FEMA USA Structures
ingest(
    'US_footprint-fema-2023',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# State parcel data.
# Change recipe_id to match the target state.
# North Carolina: US-NC_parcel-nconemap-2025
# Florida:        US-FL_parcel-floridagio-2025
ingest(
    'US-NC_parcel-nconemap-2025',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# US building points: National Structure Inventory (NSI)
ingest(
    'US_building-nsi-2022',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

```python
# US building points: National Structure Inventory (NSI)
ingest(
    'dwelling-overture-2025',
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    redownload=args.redownload,
    keep_unzipped=args.keep_unzipped,
    verbose=args.verbose,
)
```

## Harmonize

```python
harmonize(
    args.recipe_id,
    admin_ids=args.admin_ids,
    reprocess=args.reprocess,
    verbose=args.verbose,
)
```
