**`harmonize_footprints`**

Recipe-driven footprint-level building data harmonization using `Harmonizer`.

# Configure

```python
import argparse

from openplaces.io.harmonizer import Harmonizer
```

```python
parser = argparse.ArgumentParser(description='Harmonization of footprint datasets')
parser.add_argument(
    '--recipe_id',
    help='Harmonization recipe (e.g. "US_footprint-spine-2026")',
)
parser.add_argument(
    '--admin_ids',
    help='Admin unit IDs to harmonize (e.g. "US-NC-BS")',
    nargs='*',
)
parser.add_argument(
    '--reprocess',
    help='Reprocess admin IDs even if output already exists',
    action='store_true',
)
parser.add_argument(
    '--verbose',
    action='store_true',
)
```

# Test arguments

```python
ARGS_TEST = (
    '--recipe_id US_footprint-spine-2026 '
    '--admin_ids US-NC-BS '
    # '--admin_ids US-FL-AL '
    # '--admin_ids US-TX-JE '
    '--reprocess '
    '--verbose'
)

args_list = [x for x in ARGS_TEST.split(' ') if x]
args = parser.parse_args(args_list)
args
```

```python
# Show recipe parameters
from openplaces.recipe import get_recipe_by_id
from openplaces.utils import pretty_print

pretty_print(get_recipe_by_id(args.recipe_id))
```

```python
from openplaces import get_entities

get_entities('footprint-obm-2025', 'US-FL-AL')
```

```python
get_entities('US_footprint-fema-2023', 'US-FL-AL')
```

# Run

```python
harmonizer = Harmonizer(args.recipe_id, args.admin_ids, verbose=args.verbose)
```

```python
harmonizer.harmonize(reprocess=args.reprocess)
```
